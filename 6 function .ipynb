```python
#simple calculator using functions
def calculator(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        if b != 0:
            return a / b
        else:
            return "Error: Division by zero"
    else:
        return "Invalid operation"
```

```python
# write a program using fn to generate 1 to 1000
# chat gpt ka code

def perfect_number(n):
    sum = 0

    for i in range(1, n):
        if n % i == 0:
            sum = sum + i

    if sum == n:
        return True
    else:
        return False


print("Perfect numbers from 1 to 1000:")

for n in range(1, 1000):
    if perfect_number(n):
        print(n)
```

```python
# Password Strength Checker
def is_string_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()' for char in password):
        return False
    return True

##Calling the function
print(is_string_password("Abdul1@"))
print(is_string_password("Abdul"))
```

```python
# WAP to total cost of item in a shopping cart
def calculate_total_cost(Cart):
    total_cost=0
    for item in Cart:
        total_cost+=item['price']*item['quantity']

    return total_cost
cart = [
    {'name':'Apple','price':5,'quantity':4},
    {'name':'Banana','price':5,'quantity':6},
    {'name':'Orange','price':7,'quantity':3},
]

total_cost=calculate_total_cost(cart)
print(total_cost)
```
