```python
# No Argument and No Return Value

def greet():
    print("Abdul")
greet()
```

```python
# Argument and No Return Value

def greet(name):
    print("Hello", name)
greet("Abdul")
```

```python
# No Argument, with Return Value

def get_num():
    return 10
print(get_num())
```

```python
# Argument, with return value

def add(a,b):
    return a+b
print(add(2,4))
```

### Positional Argument

#### Arguments are passed in the same order as the parameters in the function definition.

```python
def students(name,age,rollno):
    print(name,age,rollno)
students("Abdul", 21, 2500320100024)
```

### Keyword Arguments
#### Explanation: Values are assigned using parameter names.

```python
def students(name,age):
    print(name,age)
students(age=21,name="Abdul")
```

```python
def palindrome(s):
    print(s.capitalize())
    print(s.lower())
    print(s.upper())
    print(s.title())
print(palindrome("a man is a good person"))
```

```python
def palindrome(s):
    s = s.lower()
    return s == s[::-1]
print(palindrome("amma"))
```
