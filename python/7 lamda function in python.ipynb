```python
# 👍
# map + lambda 
number = [1,2,3,4,5]
result = list(map(lambda x: x*2 , number))
print(result)


# filter + lambda
number = [10,20,30,40,50]
result = list(filter(lambda x:x>20 , number))
print(result)
```

```python
# higher order function
# is very Important  
names = ["alice","bob","charlee"]
scores = [86,90,34]
# combined using zip
combined = zip(names,scores)
# convert to list for diplay
print(list(combined))
```

```python
a = [1,2,3]
b = ['x','y']
print(list(zip(a,b)))
# the extra element 3 is ignored
```

```python
# unziping (reverse operation)
pairs = [('alice',85),('bob',90)]
names,scores = zip(*pairs)
print(names)
print(scores)
```

```python
print(list(zip("abc",[1,2,3],(True,False,True))))
```

```python
from itertools import zip_longest

a = [1,2,3]
b = ['x','y']
print(list(zip_longest(a,b,fillvalue='-')))
```

```python
# simple matrix multiplication using  zip function 
a = [
    [2,4],
    [4,6]
]

b = [
    [5,3],
    [3,2]
]

result = [
    [sum (a*b for a, b in zip(row,col)) for col in zip(*b)]
    for row in a
]
print(result)
```

```python
# sequence unpacking
a, *b, c ,d = [1,2,3,4,5]
print(a)
print(b)
print(c)
print(d)
```
