```python
# Comparison Operators
a = 10
b = 5
print(a>b)
print(a<b)
print(a==b)
print(a!=b)
print(a>=b)
print(a<=b)
```

```python
# Bitwise Operators
# & bitwise AND
# | bitwise OR
# ~ bitwise NOT
# ^ bitwise XOR

# >> bitwise right side
# << bitwise left side

a = 1
b = 1
print(a&b) # bitwise AND
print(a|b) # bitwise OR
print(~a)  # bitwise NOT
print(a^b) # bitwise XOR
print(a>>b)# bitwise right shift
print(a<<b)# bitwise left shift
```

```python
# Assignment Operators
#  =   +=   -=   *=   /=   %=   //=   **=   &=   |=   ^=   >>=   <<= 
x = 5 
print(x)
x += 5 
print(x)
x -= 5 
print(x)
x *= 5 
print(x)
x /= 5 
print(x)
x %= 5 
print(x)
x //= 5 
print(x)
x **= 5 
print(x)
```
