```python
marks = []

for i in range(1,6):
    mark = float(input(f"Enter marks of subject {i}: "))
    marks.append(mark)
total = sum(marks)
percentage = total/5

if percentage >= 40 and all(marks>= 33 for marks in marks):
    result = "Pass"
else:
        result = "Fail"

if percentage>=80:
    grade = "A"
elif percentage>=60:
    grade = "B"
elif percentage>=40:
    grade = "C"
else:
    grade = "D"

print("Total ", total)
print("Percentage ",percentage)
print("Result ",result)
print("Grade ",grade)
```

```python
units = float(input("Enter units consumed: "))

if units<=100:
    bill = units*5
elif units <=200:
    bill = 100*5 + (units - 100)*7
else :
    bill = 100*5 + 200*7 + (units - 200)*10

print("Bill ",bill)
```

```python
def calculate_salary(basic):
    hra = 0.20*basic
    da = 0.10*basic
    pf = 0.12*basic

    gross = basic + hra + da
    net = gross - pf
    return hra,da,pf, gross, net
basic = float(input("Enter basic salary: "))
hra, da, pf, gross, net = calculate_salary(basic)

print("Basic",basic)
print("Da",da)
print("Pf",pf)
print("Gross",gross)
print("Net",net)
```

```python
amount = float(input("Enter shopping amount: "))

if amount < 2000:
    rate = 0
elif amount <5000:
    rate = 0.05
elif amount < 10000:
    rate = 0.10
else:
    rate = 0.15

discount = amount*rate
final_price = amount - discount

print("Amount " ,amount)
print("Discount ", (discount))
print("Final Price ",final_price)
```
