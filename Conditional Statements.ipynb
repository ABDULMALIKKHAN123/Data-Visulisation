```python
#else statement
age = int(input("Enter your age:"))

if age >=18:
    print("You are allowed to vote in the election")
```

```python
# if else statement
age = int (input("Enter your age:"))
if age>=18:
    print("you are allowed to vote in the election")
else: 
    print("Your are minor and not allowed to vote in the election")
```

```python
#elif statement
age = int(input("Enter your age:"))
if age<13:
    print("You are a child")
elif age>=13 and age<20:
    print("You are a teenager")
elif age<24:
    print("You are a young adult")
```

```python
# Grade Calculator(using if, elif and else statement)
marks = int(input("Input your marks:"))
if marks>=90:
    print ("Your grade is A+")
elif marks>=80:
    print ("Your grade is A")
elif marks>=70:
    print ("Your grade is B")
elif marks>=60:
    print ("Your grade is C")
else:
    print ("Your grade is F")
```

```python
#Nested if else statement(Even or Odd)
num = int (input("Enter your Number:"))
if num>0:
    print("Number is positive")
if num%2==0:
    print("Number is even")
else:
    print ("Number is odd")
```

```python
#Nested if else statement(Leap year)
year = int(input("Enter your year:"))
if year%4==0:      
    if year%100==0:
        if year%400==0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
```

```python
#Nested calculator
num1 = float(input("Enter First Number"))
num2 = float(input("Enter Second Number"))
operation = input("Enter an operation (+, -, *, /): ")

if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    result = num1 / num2

print("The result is:", result)
```
