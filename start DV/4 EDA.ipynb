## Cleaning the Data

```python
# Remove duplicates rows
df = df.drop_duplicates()
print('Shape after removing duplicates:',df.shape)
```

```python
df.loc[df['Marks']>100,'Marks'] = np.nan
print('Max marks now:',df['Marks'].max())
```

```python
## Fill missing values with mean of the column
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].median())
print('Missing values now:')
print(df[['Marks','Age','Attendance']].isnull().sum())
print(df[['Marks','Age','Attendance']].isnull().sum().sum())
```

```python
## Standardize the Gender labels
gender_map = {'M':'Male','F':'Female'}
df['Gender'] = df['Gender'].replace(gender_map)
df['Gender'].value_counts()
```
