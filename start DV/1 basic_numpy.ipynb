## BroadCasting 
* Operation between different shape arrays
* Numpy automatically expands the smaller array (without copying) so shapes becomes compatible

```python
arr = np.array([1,2,3])  # shape (1,3)
vec = np.array([[1],[2],[3]])  # shape (3,1)
print(arr+vec)
```
