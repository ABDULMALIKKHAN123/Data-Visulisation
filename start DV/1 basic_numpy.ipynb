```python
oneD = np.array([1,2,3,4,5])

print(oneD)

# attribute of array
print(oneD.ndim)      # dimension of array
print(oneD.shape)     # shape of the array in row or column
print(oneD.size)      # number of elements in the numpy array
print(oneD.itemsize)  # size in bytes of each element 
print(oneD.dtype)     # datatype of the element
```
