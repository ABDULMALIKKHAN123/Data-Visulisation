# Rotation of Array
This is rot90() function is used to roatate an array by 90 degrees in the plane specified by axes


#### Syntax - 
  numpy.rot90(m,k=1,axes=(0,1))
  *m   Array of the or more dimensions.
  *k   Number of times the array is rotated by 90 degrees
  *axes The array is rotated in the plane defined by the axes.

```python
m = np.array([[1,2],[3,4],[6,5]])
np.rot90(m,k=3,axes=(0,1))
```
