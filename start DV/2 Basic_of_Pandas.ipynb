## Series Method
## .items Method

```python
import pandas as pd
import numpy as np
s = pd.Series(np.arange(1,10,2))
s
```

```python
for value in s:
    print(value)
```

```python
for item in s.items(): #use items()
    print(item)
```

```python
# Iteration over the index and value pairs
for idx,value in s.items():
    print(idx,value)
```

### Reset Index

* The reset_index() function is used to generate a new DataFrame or Series with the index reset.
* This is useful when the index needs to be treated as a column,or when the index is meaningless

Series.reset_index(self,level=None,drop=false,name=None,inplace=False)

```python
# Create a series
s = pd.Series([2,3,4,5],name = 'f1',index=pd.Index(['p','q','r','s'],name='idx'))
s
```

```python
# Generate a dataframe with default index
x = s.reset_index()
print(type(x))

print(type(s.reset_index()))
```

```python
s.reset_index(name="values")
```

```python
s.reset_index(drop=True)
```

```python
s
```

```python
s.reset_index(inplace=True,drop=True)
s
```
