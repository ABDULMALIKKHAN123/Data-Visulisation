## Series
### A series is similar to a single columns of data

```python
#Create a series object
s = pd.Series([145,142,24,13],name='counts')
s
```

```python
s.index
```

```python
# String as index
s = pd.Series([145,142,38,13],index=['Abdul','Malik','Khan','Majid'],name='count')
s
```

```python
s.index
```

```python
nan_ex = pd.Series([2,6,3,None])
print(nan_ex.count())
```
