## Legend

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [30, 38, 42, 51]

fig, ax = plt.subplots(figsize=(3, 3.5))
ax.plot(months, sales, marker='o', color='blue', label='Sales')
ax.set(title='Monthly Sales', ylabel='Units Sold')
ax.legend(frameon=False)
plt.show()
```

```python
import matplotlib.pyplot as plt
x = np.arange(1,6)
y = x**2
z = x**3

fig,axes = plt.subplots(1,2,figsize=(8,4))
```

```python
years = np.arange(2010,2027)
placed = [420,455,490,535,590,640]

fig, ax = plt.subplots(figsize=(7,4))
ax.plot(years, placed, marker='o', color='navy', linewidth=2,label='Placed Students')

ax.set(title='Annual Placement Trend', xlabel='Graduating Year', ylabel='Number of Students')
ax.set_xticks(years)
ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.legend(frameon=True)
plt.show()
```

```python
days = np.arange(1, 8)
temparature = [10, 15, 13, 17, 20, 18, 22]

fig,ax = plt.subplots(figsize=(6,3.4))
ax.plot(days,temparature)
ax.set(title='Weekly Temperature', xlabel='Day of the Week', ylabel='Temperature (°C)')
plt.show()
```

## Comparing two series

```python
months = ['Jan', 'Feb', 'Mar', 'Apr','May']
a = [30, 38, 42, 51, 60]
b = [25, 35, 40, 50, 55]

fig,ax = plt.subplots(figsize=(6,3.4))
ax.plot(months,a,'o-',label='Product A')
ax.plot(months,b,'s--',label='Product B')
ax.set(title='Monthly Sales', xlabel='Month', ylabel='Sales (in thousands)')
ax.legend()
plt.show()
```
