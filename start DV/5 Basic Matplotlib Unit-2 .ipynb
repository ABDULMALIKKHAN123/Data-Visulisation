## Legend

```python
months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [30, 38, 42, 51]

fig, ax = plt.subplots(figsize=(3, 3.5))
ax.plot(months, sales, marker='o', color='blue', label='Sales')
ax.set(title='Monthly Sales', ylabel='Units Sold')
ax.legend(frameon=False)
plt.show()
```

```python
years = np.arange(2010,2027)
placed = [420,455,490,535,590,640]

fig, ax = plt.subplots(figsize=(7,4))
ax.plot(years, placed, marker='o', color='navy', linewidth=2,label='Placed Students')

ax.set(title='Annual Placement Trend', xlabel='Graduating Year', ylabel='Number of Students')
ax.set_xticks(years)
ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.legend(frameon=True)
plt.show()
```

## Comparing two series

```python
months = ['Jan', 'Feb', 'Mar', 'Apr','May']
a = [30, 38, 42, 51, 60]
b = [25, 35, 40, 50, 55]

fig,ax = plt.subplots(figsize=(6,3.4))
ax.plot(months,a,'o-',label='Product A')
ax.plot(months,b,'s--',label='Product B')
ax.set(title='Monthly Sales', xlabel='Month', ylabel='Sales (in thousands)')
ax.legend()
plt.show()
```

```python
e = np.arange(1, 11)
train = [0.9, 0.85, 0.8, 0.75, 0.7, 0.65, 0.6, 0.55, 0.5, 0.45]
val = [0.88, 0.83, 0.78, 0.73, 0.68, 0.63, 0.58, 0.53, 0.48, 0.43]
best = int (np.argmin(val)) + 1

fig, ax = plt.subplots(figsize=(6,3.5))
ax.plot(e, train, 'o-', label='Training')
ax.plot(e, val, 's--', label='Validation')
ax.axvline(x=best, color='r', linestyle=':', label=f'Best Epoch: {best}')
ax.set_xlabel('Epoch')
ax.set_ylabel('Accuracy')
ax.legend()
ax.set_title('Model Performance')
```

```python
subjects = ['Python','DBMS','DSA']
a = [78, 82, 85]
b = [75, 80, 90]
x = np.arange(3)
w = 0.35

fig, ax = plt.subplots(figsize=(6,3.5))
ax.bar(x - w/2, a, width=w, label='Section A')
ax.bar(x + w/2, b, width=w, label='Section B')
ax.set_xlabel('Subjects')
ax.set_ylabel('Scores')
ax.set_title('Performance Comparison')
ax.set_xticks(x)
ax.set_xticklabels(subjects)
ax.legend()
```
